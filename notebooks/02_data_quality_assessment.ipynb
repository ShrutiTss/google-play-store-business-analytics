```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv("../data/raw/Google-Playstore.csv")
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
missing_values = (
    df.isnull()
      .sum()
      .to_frame(name="Missing Count")
)

missing_values["Missing %"] = (
    missing_values["Missing Count"] / len(df) * 100
).round(2)

missing_values = missing_values.sort_values(
    by="Missing %",
    ascending=False
)

missing_values
```

## Key Findings

- Developer Website has the highest missing percentage (32.89%).
- Privacy Policy is missing for 18.20% of apps.
- Rating-related fields have less than 1% missing data.
- Core business attributes are largely complete.

### Business Impact

The missing values are concentrated in optional metadata rather than core analytical fields, suggesting the dataset is suitable for marketplace and performance analysis after targeted cleaning.
