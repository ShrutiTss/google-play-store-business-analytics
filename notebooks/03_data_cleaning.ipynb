# Data Cleaning

```python
import pandas as pd 

df = pd.read_csv("../data/raw/Google-Playstore.csv")
```

```python
df.shape
```

### Creating a working copy:

```python
df_clean = df.copy()
```

```python
df_clean.shape
```

```python
df_clean.loc[
    df_clean["Installs"].isna(),
    ["App Name", "App Id", "Installs", "Minimum Installs", "Maximum Installs"]
]
```

```python
df_clean.loc[
    df_clean["Installs"].isna(),
    ["Minimum Installs", "Maximum Installs"]
].describe()
```

```python
# Check the rows first
df_clean[df_clean["App Name"].isna()]
```

### Observation

The missing `App Name` values do not materially impact the overall analysis, so these records will be retained.

```python
df_clean.drop(columns=["Scraped Time"], inplace=True)
```

```python
df_clean.columns
```

### Observation

`Scraped Time` was removed because it is a data collection timestamp and does not contribute to the business analysis.

## Date Conversion


```python
date_columns = ["Released", "Last Updated"]

for col in date_columns:
    df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")
```

```python
df_clean[date_columns].dtypes
```

### Action Performed

Date columns were successfully converted to datetime format.
